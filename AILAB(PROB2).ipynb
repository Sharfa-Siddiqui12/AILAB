```python
from collections import deque

def water_jug_bfs():

    start = (0, 0)
    queue = deque([(start, [])])
    visited = set()

    while queue:
        state, path = queue.popleft()

        a, b = state

        # Goal condition
        if a == 2 or b == 2:
            return path + [state]

        if state in visited:
            continue

        visited.add(state)

        next_states = []

        # Fill Jug A
        next_states.append((4, b))

        # Fill Jug B
        next_states.append((a, 3))

        # Empty Jug A
        next_states.append((0, b))

        # Empty Jug B
        next_states.append((a, 0))

        # Pour A -> B
        pour = min(a, 3 - b)
        next_states.append((a - pour, b + pour))

        # Pour B -> A
        pour = min(b, 4 - a)
        next_states.append((a + pour, b - pour))

        for next_state in next_states:
            if next_state not in visited:
                queue.append((next_state, path + [state]))

    return None


solution = water_jug_bfs()

print("Minimum steps to get 2 litres:")
for step, state in enumerate(solution):
    print("Step", step, ":", state)
```
